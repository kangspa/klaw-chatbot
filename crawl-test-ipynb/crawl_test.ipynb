```python
import time
import pandas as pd
from selenium import webdriver
from selenium.webdriver.common.by import By
```

```python
# 창 안 띄우는 옵션
options = webdriver.ChromeOptions()
options.add_argument('--headless')
# 옵션 적용 시
# driver = webdriver.Chrome(options=options)
# 옵션 미적용 시
driver = webdriver.Chrome()
```

```python
driver.get("https://www.law.go.kr/lsInfoP.do?lsiSeq=266275&efYd=20241112#0000")
```

```python
main = driver.find_element(By.CSS_SELECTOR, '#conScroll')
obj = main.text
```

```python
# 임시로 파일로 저장
file_path = '소득세법_시행령_241112.txt'
with open(file_path, 'w', encoding='utf-8') as file:
    file.write(obj)
```

```python
img_elements = driver.find_elements(By.CSS_SELECTOR, "#conScroll img")
# img 태그 배열 확인
for img in img_elements:
    # gif, png 이미지는 각각 lawicon 이미지와, 첨부파일 이미지에 해당
    src = img.get_attribute("src")
    ext = src.split(".")[-1]
    if ext not in ['gif', 'png']:
        print(src)  # 이미지 URL 출력
```
