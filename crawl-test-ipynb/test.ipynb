```python
import requests
from urllib import parse

query = "편의점에서 술을 마시는게 법적으로 허용될까요?"
encoded_query = parse.quote(query)

response = requests.get(
    "https://law.go.kr/LSW/aai/searchList.do",
    params={
        "query": encoded_query,
        "pageNum": 1,
        "pagePer": 30,
        "searchType": 1,
        "modelYn": "N",
        "keywordSearchYn": "N",
        "searchResultLsKndCd": 0,
        "searchKndCd": 0,
        "mainType": "main",
        "detailSearchYn": "N",
        "sortCode": 0
    }
)
# response = requests.get(f"https://law.go.kr/LSW/aai/searchList.do?query={query}&pageNum=1&pagePer=10&searchType=1&modelYn=N&keywordSearchYn=N&searchResultLsKndCd=0&searchKndCd=0&mainType=main&detailSearchYn=N&sortCode=0")
# print(response.text)  # 검색 결과 출력
```

```python
from bs4 import BeautifulSoup

# HTML 파싱
soup = BeautifulSoup(response.text, "html.parser")

# #itemListPC1 요소 찾기
item_list = soup.select_one("#itemListPC1")

# 요소가 존재하면 텍스트 추출
if item_list:
    text = item_list.get_text(strip=True, separator="\n")
    print(text)
else:
    print("#itemListPC1 요소를 찾을 수 없습니다.")
```

```python
import openai

# 3. OpenAI ChatGPT API에 질의
prompt = f"""당신은 한국의 법률 상담 전문 AI 어시스턴트입니다. 당신의 임무는 주어진 문맥(context)을 참고하여 질문(question) 에 답하는 것입니다.
검색된 다음 문맥(context)를 사용하여 질문(question)에 답하세요. 만약, 주어진 문맥(context)에서 답을 찾을 수 없다면, 답을 모른다면 "주어진 질문에 대한 정보를 찾을 수 없습니다."라고 답하세요.
한글로 답변하고, 답변에 참고한 문맥(context)를 답변 하단에 작성해주세요.

#Question: 
{query} 
#Context: 
{text} 

#Answer:"""

from openai import OpenAI
client = OpenAI()

response = client.responses.create(
    model="gpt-4.1-mini",
    input=prompt
)

print(response.output_text)
```
